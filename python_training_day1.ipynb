#delete instance variable


```python
class MyPython:
  def __init__(self):
    self.a = 10
    self.b = 20
    self.c = 30
    self.d = 40

  def func(self):
    del self.d

m=MyPython()
print(m.__dict__)
m.func()
print(m.__dict__)
del m.c
print(m.__dict__)
```

```python
class MyPython:
    def __init__(self):
        print("Constructor is called")

    def __del__(self):
        print("Destructor is called")

    def func(self):
        print("This is a function")

obj = MyPython()
obj.func()
del obj
```

using class & object & constructor

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(B):
    def showC(self):
        print("showC function")

obj = C()
obj.showA()
obj.showB()
obj.showC()
```

```python
class D:
    def showD(self):
        print("showD function")

class A(D):
    def showA(self):
        print("showA function")

class B(D):
    def showB(self):
        print("showB function")

class C(D):
    def showC(self):
        print("showC function")


obj1 = A()
obj1.showA()
obj1.showD()

obj2 = B()
obj2.showB()
obj2.showD()

obj3 = C()
obj3.showC()
obj3.showD()
```

```python
class Parent:
    def bike(self):
        print("Splender +")

class Child(Parent):
    def bike(self):
        print("HB")

obj = Child()
obj.bike()
```
