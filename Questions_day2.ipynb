move zeros to the end


que- given an array, move all the zeros to the end with changing the order of non zero elemnts
[0,1,0,3,12]

```python
im = [0,1,0,3,12]
for i in im:#i=0
    if i == 0:
        im.remove(i)
        im.append(i)
print(im)
```

```python
fruit_list1 = ['Apple','Berry','Cherry','Papaya']
fruit_list2 = fruit_list1
fruit_list3 = fruit_list1[:]
fruit_list2[0] = 'Guava'
fruit_list3[1] = 'Kiwi'

sum = 0
for  ls in (fruit_list1,fruit_list2,fruit_list3):
    if ls[0] == 'Guava':
        sum += 1
    if ls[1] == 'Kiwi':
        sum += 20

print(sum)
```
