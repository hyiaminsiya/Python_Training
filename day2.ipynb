```python
from abc import ABC, abstractmethod
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass
  @abstractmethod
  def m2(self):
    pass

class MyPython(Test):
  def m1(self):
    print("m1 implementation")

  @abstractmethod
  def m2(self):
    pass
class Child(MyPython):
  def m2(self):
    print("m2 implementation")

m=Child()
m.m1()
m.m2()
```

Encapsulation

```python
class Test1:
  _b=None
  __c=None
  def __init__(self):
    self.a=101
    self._b=201
    self.__c=301

  def show1(self):
    print("a= ", self.a)
    print("b= ", self._b)
    print("c= ", self.__c)

class Test2(Test1):
  def show2(self):
    print("a= ",self.a)
    print("b= ",self._b)

class Test3(Test2):
  def show3(self):
    print("a= ",self.a)
    print("b= ",self._b)

t1=Test1()
t1.show1()
t2=Test2()
t2.show2()
t3=Test3()
t3.show3()
```

```python
#we can use seek() function to move cursor (file pointed)
#specified location

data="Learning python is very easy from Ashish sir"
f=open("myfile.txt",'w')
f.write(data)

with open("myfile.txt",'r+') as f:
    fileData=f.read()
    print(fileData)
    print("Current Cursor Position is ",f.tell())
    f.seek(15)
    print("Current Cursor Position is ",f.tell())
    f.write("-JAVA-")
    f.seek(0)
    fileData=f.read()
    print(fileData)
```
